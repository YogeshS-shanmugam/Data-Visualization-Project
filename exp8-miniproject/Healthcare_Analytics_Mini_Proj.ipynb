```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
print("===== DATA TYPES BEFORE CLEANING =====")

print(df.dtypes)
```

```python
print("===== AGE ANALYSIS =====")

print("Minimum Age:", df["Age"].min())
print("Maximum Age:", df["Age"].max())

print("\nAge values below 0:")
print((df["Age"] < 0).sum())

print("\nAge values equal to 0:")
print((df["Age"] == 0).sum())
```

```python
print("===== DUPLICATE APPOINTMENT IDs =====")

duplicate_ids = df["AppointmentID"].duplicated().sum()

print("Duplicate Appointment IDs:", duplicate_ids)
```

```python
print("===== PATIENT ANALYSIS =====")

print("Total unique patients:", df["PatientId"].nunique())
print("Total appointments:", df["AppointmentID"].nunique())
```

```python
# Convert date columns to datetime format

df["ScheduledDay"] = pd.to_datetime(df["ScheduledDay"])
df["AppointmentDay"] = pd.to_datetime(df["AppointmentDay"])

print("ScheduledDay type:", df["ScheduledDay"].dtype)
print("AppointmentDay type:", df["AppointmentDay"].dtype)
```

```python
# Calculate waiting time in days

df["Waiting_Days"] = (
    df["AppointmentDay"].dt.normalize()
    - df["ScheduledDay"].dt.normalize()
).dt.days

print("===== WAITING DAYS =====")

print(df["Waiting_Days"].describe())
```

```python
print("Negative waiting days:", (df["Waiting_Days"] < 0).sum())
```

```python
df["Attendance_Status"] = df["No-show"].map({
    "No": "Attended",
    "Yes": "No-Show"
})

print(df["Attendance_Status"].value_counts())
```

```python
def create_age_group(age):
    if age < 18:
        return "0-17"
    elif age <= 30:
        return "18-30"
    elif age <= 45:
        return "31-45"
    elif age <= 60:
        return "46-60"
    elif age <= 75:
        return "61-75"
    else:
        return "76+"

df["Age_Group"] = df["Age"].apply(create_age_group)

print("===== AGE GROUP DISTRIBUTION =====")

print(df["Age_Group"].value_counts())
```

```python
def create_waiting_group(days):
    if days <= 0:
        return "Same Day"
    elif days <= 7:
        return "1-7 Days"
    elif days <= 14:
        return "8-14 Days"
    elif days <= 30:
        return "15-30 Days"
    elif days <= 60:
        return "31-60 Days"
    else:
        return "60+ Days"

df["Waiting_Time_Group"] = df["Waiting_Days"].apply(create_waiting_group)

print("===== WAITING TIME GROUP =====")

print(df["Waiting_Time_Group"].value_counts())
```

```python
df["Health_Conditions"] = (
    df["Hipertension"]
    + df["Diabetes"]
    + df["Alcoholism"]
)

print("===== HEALTH CONDITION COUNT =====")

print(df["Health_Conditions"].value_counts().sort_index())
```

```python
df["SMS_Status"] = df["SMS_received"].map({
    0: "No SMS",
    1: "SMS Received"
})

print("===== SMS STATUS =====")

print(df["SMS_Status"].value_counts())
```

```python
print("===== UPDATED DATASET =====")

print("Rows:", df.shape[0])
print("Columns:", df.shape[1])

print("\nNew Columns:")

print([
    "Waiting_Days",
    "Attendance_Status",
    "Age_Group",
    "Waiting_Time_Group",
    "Health_Conditions",
    "SMS_Status"
])

print("\nFirst 5 rows:")

df.head()
```

```python
# Save cleaned dataset

df.to_csv(
    "Healthcare_Appointment_Cleaned.csv",
    index=False
)

print("Cleaned dataset saved successfully!")
```

```python
# Top 10 Neighbourhoods by No-Show Rate

neighbourhood_rate = (
    df.groupby("Neighbourhood")["No-show"]
    .apply(lambda x: (x == "Yes").mean() * 100)
    .sort_values(ascending=False)
)

top_10_neighbourhoods = neighbourhood_rate.head(10)

print("===== TOP 10 NEIGHBOURHOODS BY NO-SHOW RATE =====")

print(top_10_neighbourhoods)

plt.figure(figsize=(10, 6))

plt.barh(
    top_10_neighbourhoods.index[::-1],
    top_10_neighbourhoods.values[::-1]
)

plt.title("Top 10 Neighbourhoods by No-Show Rate")
plt.xlabel("No-Show Rate (%)")
plt.ylabel("Neighbourhood")

plt.show()
```
