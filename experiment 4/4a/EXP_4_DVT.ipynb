```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("KaggleV2-May-2016.csv")
```

```python
print("===== FIRST FIVE RECORDS =====")
display(df.head())
```

```python
print("\n===== SHAPE OF DATASET =====")
print(df.shape)
```

```python
print("\n===== SHAPE OF DATASET =====")
print(df.shape)
```

```python
print("\n===== SUMMARY =====")
display(df.describe(include="all"))
```

```python
print("\n===== PATIENTS WITH AGE > 60 =====")
display(df[df["Age"] > 60].head())
```

```python
print("\n===== FEMALE PATIENTS =====")
display(df[df["Gender"] == "F"].head())
```

```python
print("\n===== NO SHOW PATIENTS =====")
display(df[df["No-show"] == "Yes"].head())
```

```python
print("\n===== SORT BY AGE (DESCENDING) =====")
display(df.sort_values(by="Age", ascending=False).head())
```

```python
print("\n===== AGE STATISTICS =====")
print("Mean :", df["Age"].mean())
print("Median :", df["Age"].median())
print("Mode :")
print(df["Age"].mode())
print("Standard Deviation :", df["Age"].std())
```

```python
print("\n===== DESCRIPTIVE STATISTICS =====")
display(df.describe())
```

```python
print("\n===== AVERAGE AGE BY GENDER =====")
gender_avg = df.groupby("Gender")["Age"].mean()
display(gender_avg)
```

```python
print("\n===== CORRELATION MATRIX =====")

corr = df[[
    "Age",
    "Scholarship",
    "Hipertension",
    "Diabetes",
    "Alcoholism",
    "Handcap",
    "SMS_received"
]].corr()

display(corr)
```

```python
plt.figure(figsize=(5,4))
gender_avg.plot(kind="bar")
plt.title("Average Age by Gender")
plt.xlabel("Gender")
plt.ylabel("Average Age")
plt.show()
```

```python
plt.figure(figsize=(8,5))
plt.scatter(df["Age"], df["SMS_received"])
plt.title("Age vs SMS Received")
plt.xlabel("Age")
plt.ylabel("SMS Received")
plt.show()
```

```python
plt.figure(figsize=(8,5))

plt.hist(df["Age"], bins=20)

plt.title("Age Distribution")
plt.xlabel("Age")
plt.ylabel("Frequency")
plt.show()
```
