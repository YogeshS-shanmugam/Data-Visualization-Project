```python
import pandas as pd
import matplotlib.pyplot as plt

# Import Dataset
df = pd.read_csv("hotel_bookings.csv")


# Dataset Inspection
print("===== FIRST FIVE RECORDS =====")
display(df.head())

print("\n===== SHAPE OF DATASET =====")
print(df.shape)

print("\n===== DATA TYPES =====")
print(df.dtypes)

print("\n===== SUMMARY =====")
display(df.describe())


# Filtering Data

# Bookings with ADR greater than 100
print("\n===== BOOKINGS WITH ADR > 100 =====")
high_adr = df[df["adr"] > 100]
display(high_adr.head())

# Resort Hotel bookings
print("\n===== RESORT HOTEL BOOKINGS =====")
resort = df[df["hotel"] == "Resort Hotel"]
display(resort.head())

# Sorting Data
print("\n===== SORT BY ADR (DESCENDING) =====")
sorted_df = df.sort_values(by="adr", ascending=False)
display(sorted_df.head())

# Statistical Analysis
print("\n===== ADR STATISTICS =====")
print("Mean :", df["adr"].mean())
print("Median :", df["adr"].median())
print("Mode :")
print(df["adr"].mode())
print("Standard Deviation :", df["adr"].std())

print("\n===== DESCRIPTIVE STATISTICS =====")
display(df.describe())

# Grouping and Aggregation
print("\n===== AVERAGE ADR BY HOTEL TYPE =====")
hotel_avg = df.groupby("hotel")["adr"].mean()
display(hotel_avg)

# Correlation Analysis
print("\n===== CORRELATION =====")
corr = df[[
    "lead_time",
    "stays_in_weekend_nights",
    "stays_in_week_nights",
    "adults",
    "children",
    "babies",
    "adr",
    "is_canceled"
]].corr()

display(corr)

# Visualization 1
hotel_avg.plot(kind="bar")
plt.title("Average ADR by Hotel Type")
plt.xlabel("Hotel")
plt.ylabel("Average ADR")
plt.show()

# Visualization 2
plt.figure(figsize=(8,5))
plt.scatter(df["lead_time"], df["adr"])
plt.title("Lead Time vs ADR")
plt.xlabel("Lead Time")
plt.ylabel("ADR")
plt.show()
```
