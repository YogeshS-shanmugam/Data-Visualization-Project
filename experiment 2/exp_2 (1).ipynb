```python

# EDA - Data Import and Export


# Import required libraries
import pandas as pd
import sqlite3

# Load the Hotel Booking dataset
df = pd.read_csv("/content/hotel_bookings.csv")

# Display the complete dataset
print("===== DATASET =====")
display(df)

# Display first five records
print("\n===== FIRST FIVE ROWS =====")
display(df.head())

# Display last five records
print("\n===== LAST FIVE ROWS =====")
display(df.tail())

# Display dataset information
print("\n===== DATASET INFORMATION =====")
df.info()

# Display statistical summary
print("\n===== STATISTICAL SUMMARY =====")
display(df.describe(include='all'))

# Display dataset shape
print("\nDataset Shape:")
print(df.shape)

# Display column names
print("\nColumns:")
print(df.columns)

# Check missing values
print("\n===== MISSING VALUES =====")
print(df.isnull().sum())

# Export dataset to CSV
df.to_csv("hotel_bookings_export.csv", index=False)
print("\nCSV file exported successfully!")

# Export dataset to Excel
df.to_excel("hotel_bookings_export.xlsx", index=False)
print("Excel file exported successfully!")

# Create SQLite database
conn = sqlite3.connect("hotel_booking.db")

# Export DataFrame to SQL table
df.to_sql(
    "hotel_bookings",
    conn,
    if_exists="replace",
    index=False
)

print("Dataset exported successfully to SQLite database!")

# Close the connection
conn.close()
```

```python
import pandas as pd

# Load dataset, skipping bad lines to avoid ParserError
df = pd.read_csv("/content/sample_data/hotel_bookings.csv", on_bad_lines='skip', engine='python')

print("===== DATASET INFO ====")
df.info()

print("\n===== MISSING VALUES IN EACH COLUMN ====")
print(df.isnull().sum())

print("\n===== TOTAL MISSING VALUES ====")
print(df.isnull().sum().sum())

print("\n===== ANY MISSING VALUES? ====")
print(df.isnull().values.any())

print("\n===== NON-MISSING VALUES COUNT ====")
print(df.notnull().sum())
```

```python

# Ensure 'children' column is numeric, coercing errors to NaN
df['children'] = pd.to_numeric(df['children'], errors='coerce')

# Fill all missing values with 0
df_constant = df.fillna(0)
print("===== CONSTANT VALUE ====")
display(df_constant.head())

# Fill 'children' column with mean
df_mean = df.copy()
df_mean["children"] = df_mean["children"].fillna(df_mean["children"].mean())
print("\n===== MEAN ====")
display(df_mean.head())

# Fill 'children' column with median
df_median = df.copy()
df_median["children"] = df_median["children"].fillna(df_median["children"].median())
print("\n===== MEDIAN ====")
display(df_median.head())

# Fill 'country' column with mode
df_mode = df.copy()
df_mode["country"] = df_mode["country"].fillna(df_mode["country"].mode()[0])
print("\n===== MODE ====")
display(df_mode.head())

# Forward Fill
df_ffill = df.ffill()
print("\n===== FORWARD FILL ====")
display(df_ffill.head())

# Backward Fill
df_bfill = df.bfill()
print("\n===== BACKWARD FILL ====")
display(df_bfill.head())
```
